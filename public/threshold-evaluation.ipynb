# Decision threshold evaluation
Place threshold-data.csv alongside this notebook. The 16 rows are synthetic.

```python
import csv
from pathlib import Path

threshold = 0.5
with Path("threshold-data.csv").open() as source:
    pairs = [(float(row["score"]) >= threshold, int(row["label"])) for row in csv.DictReader(source)]
tp = sum(pred and label for pred, label in pairs)
fp = sum(pred and not label for pred, label in pairs)
fn = sum(not pred and label for pred, label in pairs)
tn = sum(not pred and not label for pred, label in pairs)
precision = tp / (tp + fp) if tp + fp else None
recall = tp / (tp + fn) if tp + fn else None
print(dict(tp=tp, fp=fp, tn=tn, fn=fn, precision=precision, recall=recall))
```
